#LIST

* To Store and Retrive multiple values under the single variable name

* Must be enclosed with []
* index based - It starts from 0 (zero)
* Mutable - add ,remove ,update
* It is Ordered
* It allow all the Datatypes
* It allow the Duplicate


```python
anewlist = ['naveen','niranjan','pravin','saranya','keerthana','barath','prasanth','Saranya','saranya','keerthana','keerthana','varun']

print(anewlist)
print (len (anewlist))
print(anewlist[4])
print(anewlist[0])

#Mutability

anewlist.append('venkataramana') #adding the data
print(anewlist)

print('----'*15)

anewlist.remove('saranya') #remove the data
print(anewlist)

anewlist.remove(anewlist[2])  # delete using index
print(anewlist)

print('----'*15)

anewlist.reverse() #reverse the data
print(anewlist)

print('----'*15)

#Updating
anewlist.insert(1,'Deepan') # update the data
print(anewlist)

print('----'*15)

anewlist.pop(4) # delete using index
print(anewlist)

anewlist.extend(['Deepan']) # adding name at last
print(anewlist)



```

* It allows all the datatypes

#SET
* It is not at all ordered,There is no indexing
* It wont allow dulicates
* it allows all the datatypes


```python
myset = {'naveen','niranjan','pravin','saranya','keerthana','barath','prasanth','Saranya','saranya','keerthana','keerthana','varun'}
print(myset)
print(len(myset))
myset.add('Deepan') #add new name
print(myset)
myset.remove('saranya') # remove the name
print(myset)

setone = {33,25,46,89,95}
setwo = {38,46,95,28,49}

union_function = setone.union(setwo) # join the both set funtion
print(union_function)

intersectfunction = setone.intersection(setwo) # same value present in both set
print(intersectfunction)

differencefunction = setone.difference(setwo) # not similiar function in first set
print(differencefunction)

symmetricdifferencefunction = setone.symmetric_difference(setwo) # Not similiar function in second set
print(symmetricdifferencefunction)

myset.discard('naveen') # remove the items that exists
print(myset)

myset.clear() # clear the set
print(myset,'set clear')
```

# DICTIONARY

*   To store and retrieve data in {key-value} pairs.
*   Must be enclosed with `{}`.
*   Inside the dictionary u can create multiple in built datatypes (list,tuple,set,dictionary)
*   Keys must be unique and immutable (e.g., strings, numbers, tuples).
*   Values can be of any data type and can be duplicated.
*   Mutable - add, remove, update elements.
*   Unordered (in Python versions before 3.7, ordered from 3.7 onwards).
*   Access elements using keys, not numerical indices.

```python
mydata = {
    'fname':'Niranjan',
    'sname':'subbaraya',
    'age' : 28,
    'location' :'chennai',
    'profession' :['AI Engineer','Trainer']
}
print(mydata)
print(type(mydata))
print(len(mydata))
print(mydata['fname'])
print(mydata['profession'])
print(mydata['profession'][1])
print(mydata.keys())
print(mydata.values())
print(mydata.items())

print('-------'*10)

individualData = mydata.get('fname')
print(individualData)

print('-------'*10)

# Accessing values using keys
print(mydata.get('sname'))
print(mydata.get('profession'))

mydata['location'] = 'banglore'
print(mydata)

mydata['profession'][0] = 'data science'
print(mydata)




```

```python

my_dict = {
    'name': 'naveen',
    'age': 30,
    'city': 'Chennai',
    'occupation': 'Engineer',
    'skills': ['Python', 'SQL', 'Data Analysis']
}
print("Original dictionary:", my_dict)
print("Type of my_dict:", type(my_dict))
print("Length of my_dict:", len(my_dict))

print('\n--- Accessing elements ---\n')
# Accessing values using keys
print("Name:", my_dict['name'])
print("Age:", my_dict['age'])
print("Skills:", my_dict['skills'])

# Using .get() method (safer, returns None if key not found)
print("City (using .get()):", my_dict.get('city'))
print("Country (using .get()):", my_dict.get('country')) # Key 'country' does not exist

print('\n--- Modifying elements ---\n')
# Updating a value
my_dict['age'] = 31
print("Updated age:", my_dict['age'])

# Adding a new key-value pair
my_dict['email'] = 'naveen@example.com'
print("Dictionary after adding email:", my_dict)

print('\n--- Removing elements ---\n')
# Removing a key-value pair using del
del my_dict['occupation']
print("Dictionary after deleting occupation:", my_dict)

# Removing a key-value pair using .pop()
removed_city = my_dict.pop('city')
print("Removed city:", removed_city)
print("Dictionary after popping city:", my_dict)

# Removing the last inserted item (for Python 3.7+ dictionaries, order is insertion order)
# Note: Behavior of popitem() is arbitrary before Python 3.7
my_dict['new_key'] = 'new_value'
last_item = my_dict.popitem()
print("Popped last item:", last_item)
print("Dictionary after popitem():", my_dict)

# Clearing all elements
my_dict.clear()
print("Dictionary after clear():", my_dict)
```
