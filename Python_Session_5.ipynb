#Session 5 - Loop
 **For loop**

* A for loop in Python is a control flow statement used to iterate over a sequence

* (like a list, tuple, dictionary, set, or string) and execute a block of code once for each item in that sequence.

* Unlike traditional for loops in other languages, Python's version functions.

* like an iterator method, directly pulling elements from a collection without requiring you to manually manage loop indexes.

**For item in sequence: This line starts the loop
* *`Item`* is a variable that takes on the value of each element in the sequence during each iteration
* *`sequence`* is the collection of items you want to loop through
* The colon(:) is crucial; it indicates start of the loop's code block.
**Indentation** : The code within the loop block must bee indented. Python uses indentation to define code blocks

**Looping with `Enumerate()`:**
enumerate() adds a counter to an iterable.It's useful when you need both the index and the valve of an item.

**Examples**

```python
inventory ={
    'laptop':[999.99,1],
    'smartphone':[499.99,5],
    'headphones':[89.99,1],
    'keyboard' :[45.00,15]
}
# A shopping cart containing (item_name, Quantity_reqested)
shopping_cart = [
    ('laptop',1),
    ('headphones',1),
    ('keyboard',2),
    ('smartphone',3)
]
# 2. Automated Restocking(looping through a dictionary)
print('---Step 1: Checking and Restocking low Inventory---')
for item,details in inventory.items():
    current_stock = details[1]
    if current_stock < 5:
        print(f'low stock alert for {item} ({current_stock}left).Addimg 10 units.')
        inventory[item][1] += 10
# 3. Processing the shopping cart (looping through a list of tuples)

        print('---Step 2: Processing Customer Order---')
        subtotal = 0
        processed_items =[]
        for item_name, qty in shopping_cart:
          if item_name in inventory:
            price = inventory[item_name][0]
            available_stock = inventory[item_name][1]

            if available_stock >= qty:
              inventory[item_name][1] -=qty
              item_total = price * qty
              subtotal += item_total
              processed_items.append ((item_name ,qty ,item_total))
              print(f'Added {qty}x {item_name} to order. Total: ${item_total:.2f}')
            else:
                print(f'connot fulfill {qty}x {item_name}.Only {available_stock} available.')
          else:
            print(f'{item_name} is not carried in this store.')

discount = 0.0
if subtotal > 1000:
  discount = subtotal * 0.10

tax_rate = 0.08
total_tax = (subtotal-discount) * tax_rate
final_total = (subtotal-discount) + total_tax

# 5.GENERATING THE INVOICE(LOOPING WITH INDEX ENMERRATION)
print('================================')
print('         FINAL INVOICE            ')
print('==================================')
for index,(item, qty, cost) in enumerate(processed_items, start=1):
  print(f'{index}.{item:<12} x{qty} -> ${cost:>8.2f}')
print('==================================')
print(f'Subtotal:      ${subtotal:>8.2f}')
print(f'10% Discount: -${discount:>8.2f}')
print(f'Tax(8%):       ${total_tax:>8.2f}')
print(f'Grand Total:   ${final_total:>8.2f}')
print('==================================')
```

#**Using range():**
  The range () function generates a sequence of numbers.It's very comman to use it with `for loops`.

  * `range(stop)` : Generates numbers from 0 up to (but not including). `stop`

  * `range(start, stop) `: Generates numbers from start up to (but not including) `stop`

  * `range(start, stop ,step)` : Generates numbers from start up to (but not including) stop,incrementing by `step`

```python
# range funtion (Start,Stop,Step)

for numbers in range (10):
 print(numbers)

for numbers in range (1,20):
  print(numbers,'start at one and end at 20')

for numbers in range (0,20,5):
  print(numbers)
```
