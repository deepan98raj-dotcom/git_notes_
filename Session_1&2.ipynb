```python


age = 27
has_license = True

if age >= 18 and has_license:
  print('you are allowed to drive.')
else:
  print('you are not allowed to drive.')
```

```python
score = 95
#Check if the score is between 80 and 90(inclusive)
if score >= 90 and score <= 100:
  print('Grade-A+')
elif score >= 80 and score <= 90:
  print('GRADE-A')
elif score >= 70 and score < 80:
  print('GRADE-B')
elif score >= 60 and score < 70:
  print('GRADE-C')
elif score >= 40 and score < 60:
  print('GRADE-D')
elif score >= 0 and score < 40:
  print('GRADE-F')
else :
  print('Invalid score')

```

```python
#OR Operations

# Even though 10/0 would cause an error , it is never executed
# Because 5 > 1 is already true
if 5 > 1 or [(10/0) > 1]:
  print('This code runs safely.')
```

```python
user_choice = input ('enter your answer (yes/no):').lower()

# Check if the user entered Any Variation of 'yes'
if user_choice =="yes" or user_choice == "y":
  print ("action is confirmed")
elif user_choice == "no" or user_choice =="n":
  print ("action cancelled")
else:
    print("invalid input")

```
