```python
mnist = tf.keras.datasets.mnist
mnist
(x_train, y_train),(x_test,y_test) = mnist.load_data()
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Flatten, Input
```

Sequential : it acts as our model containers.all layers can be stacked one by one . first input, Next hidden layers, output layers

Dense : All neurons are interconnected with each other which is also called as fully Connected layers

Flatten : It is an important layer , NN doesn't understand Matrix, 28*28 = 784 pixels will be converted as a line.

Input : Mentioning the size of the input data, 28*28

Dropout : To avoid Overfitting, It OFF certain neurons in the dataset

```python
model = Sequential()
model.add(Input(shape=(28,28)))
model.add(Flatten())
model.add(Dense(128,activation='relu'))
model.add(Dropout(0.2))
model.add(Dense(10))
```

```python
general_prediction = model(x_train[:1]).numpy()
general_prediction
```

```python
tf.nn.softmax(general_prediction).numpy().sum()
```

```python
from tensorflow.keras.losses import SparseCategoricalCrossentropy
loss_fn = SparseCategoricalCrossentropy(from_logits = True)

-tf.math.log(1/10)
```

```python
loss_fn(y_train[:1],general_prediction)
```

```python
from tensorflow.keras.optimizers import Adam

optimizer = Adam(learning_rate=0.01)
model.compile(optimizer='adam', loss=loss_fn, metrics=['accuracy'])
```

```python

model.fit(x_train, y_train, epochs=15, validation_data=(x_test, y_test))
```
