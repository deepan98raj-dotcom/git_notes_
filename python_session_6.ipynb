```python
def greet_person(gender_indicator, person_name):
    # The if/elif/else statements need colons and proper indentation.
    # 'input' is a function, not a variable to check against.
    # We should use the function parameters (like gender_indicator) for comparison.
    # The f-strings need to be prefixed with 'f'.
    if gender_indicator.lower() == 'm':
        print(f'\n welcome to Gen AI Mr. {person_name}')
    elif gender_indicator.lower() == 'f':
        print(f'\n welcome to Gen AI Ms. {person_name}')
    else:
        print(f'\n welcome to Gen AI {person_name}')

# Example usage:
greet_person('m', 'Deepan')
greet_person('f', 'Alice')
greet_person('other', 'Alex')
```

```python
def welcome_candidate(name, course):
    # Normalize the text so capitalization doesn't break the logic
    course_lower = course.lower()

    # Check which course they are joining using if-elif-else
    if course_lower == "python":
        print(f"Welcome to the team, {name}! 🚀 Get ready to master functions, loops, and logic in our Python Course.")
    elif course_lower == "gen ai" or course_lower == "ai":
        print(f"Welcome to the team, {name}! 🤖 Step into the future and build amazing things in our Gen AI Course.")
    else:
        # Default fallback message if the course doesn't match the ones above
        print(f"Welcome to the team, {name}! ✨ We are thrilled to have you join the {course} course.")

# --- Test the function with different candidates ---
welcome_candidate("Deepan", "Python")
welcome_candidate("Alice", "Gen AI")
welcome_candidate("Natesh", "Data Science")
```

```python
def check_enrollment_status(name, course, is_paid):
    # Using if-else to check the payment status
    if is_paid:
        print(f"🎉 Welcome to the class, {name}! Your enrollment for {course} is complete. See you in the session!")
    else:
        print(f"👋 Welcome, {name}! You have successfully registered for {course}, but your payment is pending.")

        # A quick while loop example to simulate waiting for confirmation
        reminder_count = 1
        while reminder_count <= 2:
            print(f"   [Reminder {reminder_count}]: Please complete your fee submission to unlock the course materials.")
            reminder_count += 1

# --- Testing the function ---
print("--- Candidate 1 ---")
check_enrollment_status("Deepan", "Python", is_paid=True)

print("\n--- Candidate 2 ---")
check_enrollment_status("Alice", "Gen AI", is_paid=False)
```

```python
def process_new_candidate():
    print("--- Welcome to the Course Enrollment Portal ---")

    # 1. Using input() and type conversion to collect details
    name = input("Enter candidate's name: ")
    age = int(input("Enter candidate's age: "))
    course = input("Enter course name (Python / Gen AI): ").lower()

    # 2. if-elif-else logic to check course and age eligibility
    if course == "python":
        if age >= 18:
            print(f"\n🎉 Welcome {name}! You are successfully enrolled in the Python Course.")
        else:
            print(f"\n👋 Welcome {name}! You're enrolled in Python Kids Edition due to your age ({age}).")

    elif course == "gen ai" or course == "ai":
        # Nested if-else for age-based Gen AI enrollment
        if age >= 18:
            print(f"\n🤖 Welcome {name}! Get ready to explore Generative AI.")
        else:
            print(f"\n👋 Welcome {name}! You're enrolled in GEN AI Kids Edition due to your age ({age}).")
    else:
        print(f"\n✨ Welcome {name}! You have selected the {course.title()} course.")

    # 3. Integrating a while loop to handle batch setup (e.g., creating 3 profile setup steps)
    step = 1
    print("\nSetting up your student account portal...")
    while step <= 3:
        print(f"   [Step {step} of 3]: Configuring setup parameters...")
        step += 1

    print("\n✅ Registration Complete! Happy Learning!")

# --- Call the function to run the example ---
process_new_candidate()
```
