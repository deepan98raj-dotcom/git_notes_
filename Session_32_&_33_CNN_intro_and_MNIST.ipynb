**What is a CNN?**

In simple terms, a CNN is **giving the computer the power of sight**.

Just as our eyes look at an object and our brain identifies it as a cat or a dog, a CNN is a special type of neural network that allows computers to **SEE** and understand images.

**How does a CNN work?**

Imagine we show a computer a photo of a cat. How does it know it's a cat? A CNN works in 4 main stages:

**Convolution Layer** also called as **Feature Extraction**:

The computer doesn't look at the whole photo at once. It "scans" it in small pieces to find specific features.

Are there ears? Are there eyes? Is there a mustache?

It uses a "Filter" (or Kernel) that moves over the photo to detect edges, curves, and colors.

**ReLU Layer (Activation)**
The computer only cares about "positive" features. The ReLU (Rectified Linear Unit) function converts negative values to 0. This ensures that only the "active" features move forward to the next stage.

**Pooling Layer (Down-sampling)**
If a photo is too large, it takes too much processing power. The Pooling layer keeps the most important information and removes unnecessary pixels.

**Example:**
Max Pooling looks at a small area and keeps only the highest (brightest) value.

**Fully Connected Layer (Classification)**

Now, all the features are ready.

Ears found ?,  Mustache found? ,Face shape matches? **Yes.**

Finally, the computer combines these facts and says: There is a 95% chance this is a cat!

**A Simple Example:**

Think of a Jigsaw Puzzle:

**Convolution:** You separate the pieces and look at the colors and edges.

**Pooling:** You throw away the background pieces that don't help you build the main image.

**Fully Connected:** You put the pieces together, look at the finished picture, and say, "Oh, it's a cat!"

**Why do we use CNNs?**

Normal neural networks look at every single pixel, which takes too much time. CNNs are better because:

Space Efficient: They focus only on the important features.

Efficient: They can recognize an object even if it's moved or shifted in the photo.

**Practical Applications:**

Face Recognition: The facial lock that unlocks your phone.

Self-driving Cars: Detecting road signs and pedestrians.

Medical Imaging: Identifying diseases in X-rays or MRI scans.

Image → Convolution → Pooling → Flatten → Dense → Sigmoid → Cat or Dog

This code prepares the MNIST dataset (handwritten digits 0–9) for your AI model.

Imports: We bring in TensorFlow to build the model, layers/models to define the architecture, and matplotlib to visualize the images.

mnist.load_data(): This function does three things automatically:

Downloads the MNIST dataset (60,000 training images + 10,000 testing images).

Converts them into computer-readable numbers (NumPy arrays).

Splits them into Training data (for the model to learn) and Testing data (for the model to take its "final exam").

```python
import tensorflow as tf
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt

# 1. Dataset loading
mnist = tf.keras.datasets.mnist
(train_images, train_labels),(test_images, test_labels) = mnist.load_data()
```

This code prepares the data specifically for a CNN (which expects 2D images, not flat lists):

**.reshape((..., 28, 28, 1)):**

CNNs need to see the image as a grid. This adds an extra dimension at the end (1), representing the grayscale channel. Where CNN needs 3rd dimension color channel

It tells the model: This is a 28x28 image with 1 color channel.

**.astype('float32'):**

Converts pixel values from integers (0–255) to floating-point numbers. This makes the math (matrix multiplication) faster and more precise.

**/ 255:**
Normalization. Pixel values range from 0 to 255. Dividing by 255 squeezes them into a 0 to 1 range.

Neural networks learn much faster and more stably with small numbers (0–1) than with large ones (0–255).

In short: You are formatting the images into a 3D grid and shrinking the pixel values to 0–1 to help the CNN learn faster and more accurately.

```python
# 2. Preprocessing (Normalization & Reshaping)

train_images = train_images.reshape((60000, 28, 28, 1)).astype('float32') / 255
train_images = train_images.reshape((60000, 28, 28, 1)).astype('float32') / 255
```

Convolutional Base **Feature Extraction**
This is where the AI sees at the image:

**Conv2D (32, 64):** These filters act like scanners to find patterns in the image, such as edges, curves, and lines.

**MaxPooling2D (2, 2):** This reduces the image size by half (down-sampling). It makes the computer's job easier by removing unnecessary details while keeping the most important features.

**Flatten**
Flatten(): This converts the 2D grid data from the previous layers into one long, 1D list. This is necessary because the final Dense layers can only understand a simple list of numbers, not a grid.

**Classification acts as the Brain**:

**Dense(64, 'relu'):** The model uses the features it found to learn complex patterns and relationships.

**Dense(10, 'softmax'):** This is the final layer. There are 10 neurons because there are 10 possible digits (0–9) in the MNIST dataset.

**Softmax function:** It calculates the probability (from 0% to 100%) for each of the 10 classes, telling you which digit is the most likely match.

```python
# 3. CNN Architecture
model = models.Sequential([
# Convolutional Base
layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
layers.MaxPooling2D((2, 2)),

layers.Conv2D(64, (3, 3), activation='relu'),
layers.MaxPooling2D((2, 2)),

layers.Conv2D(64, (3, 3), activation='relu'),

# Flatten & Dense Layers (Classification)
layers.Flatten(),
layers.Dense(64, activation='relu'),
layers.Dense(10, activation='softmax')
])
```

This is the setup phase where you tell the model how to learn and how to measure its own success. Here is the simple breakdown:

**optimizer='adam'**
What it does: Think of this as the "Teacher" or the "Learning Strategy."

**Simple Explanation:** After every guess the model makes, it compares the answer with the correct one. The adam optimizer is the algorithm that adjusts the model’s internal settings (weights) to make better guesses next time. It is the most popular choice because it learns quickly and efficiently.

**loss='sparse_categorical_crossentropy'**
What it does: This is the Penalty System.

**Simple Explanation:** This function measures how wrong the model's guess was. If the model is very confident that a '7' is actually a '2', the loss value will be high (a big penalty). The model's goal is to make this penalty as close to zero as possible.

Note: We use sparse because our labels (0–9) are simple integers, not complicated lists.

**metrics=['accuracy']**

**What it does:** This is the "Report Card."

**Simple Explanation:** This is just for us (the humans) to see. It tracks the percentage of correctly classified images during training. It doesn't affect how the model learns; it just helps us understand how well it is performing.

```python
# 4. Compile & Train
model.compile(optimizer= 'adam',
              loss='sparse_categorical_crossentropy',
              metrics = ['accuracy'])
```

```python
# Model summary
model.summary()
# Training
history = model.fit(train_images, train_labels, epochs=5, validation_split=0.1)

# 5.Evaluate
test_loss, test_acc = model.evaluate(test_images, test_labels)
print(f"\nFinal Test accuracy: {test_acc*100:.2f}%")
```

```python
import tensorflow as tf

#1 Download & Extract
dataset_url = "https://storage.googleapis.com/download.tensorflow.org/example_images/flower_photos.tgz"
data_dir = tf.keras.utils.get_file('flower_photos',origin=dataset_url,untar=True)

#2. Prints the path and Checks
print(f'Dataset path: {data_dir}')

#3. Load dataset
batch_size = 32
img_height = 180
img_width = 180

train_ds = tf.keras.utils.image_dataset_from_directory(
    data_dir,
    validation_split=0.2,
    subset="training",
    seed=123,
    image_size=(img_height, img_width),
    batch_size=batch_size)

val_ds = tf.keras.utils.image_dataset_from_directory(
    data_dir,
    validation_split=0.2,
    subset="validation",
    seed=123,
    image_size=(img_height, img_width),
    batch_size=batch_size)

# Normalization
normalization_layer = tf.keras.layers.Rescaling(1./255)
train_ds = train_ds.map(lambda x, y: (normalization_layer(x), y))
val_ds = val_ds.map(lambda x, y: (normalization_layer(x),y))

print('Data loaded sucessfully')
```
