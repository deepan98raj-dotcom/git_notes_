# **Old - Style String Formatting (Using the % Operator)**
 This is the oldest method , reminiscent of C's printf style. while still functional , it's generally considered less readable and flexible compared to newer methods.

*   Syntax : '%[flags] [width] [.precision]conversion_type' % values
*   Conversion Types  :                                                         
        1.   %s : String (or any object with a string representation)                                                         
        2.   %d : Signed decimal integer                                        
        3.   %f : Floating-point decimal                                        
        4.   %x/%X : Hexadecimal integer (lowercase/uppercase)                  
        5.   %o : octal integer                                                 
        6.   %% : literal percent sign

#2 str.format() Method (Modern String Formatting)

 This method is more versatile and readable then % operator

## 1.   Syntax:
           '{}'.format (values) or '{field_name}'.format(values)
## 2.   Features
        * positional arguments:'{0},{1},{2}'.format(a,b,c)
        * Keyword arguments:'{name},{age}'.format{name=alice',age = 30}
        * accessing attributes: '{person.name}, {person.age}'.format{person=person('alice',30))
        * Formatting specifiers:'{:.2f}'.format(3.14159)

```python
class person:
  #
  def __init__(self,name,age,location):
    self.name = name
    self.age = age
    self.location = location

  def greet (self):
    return f'hi my name is {self.name} and i am {self.age} years old . and i am residing in {self.location}'

#creating objects
personone = person('nivethitha',23,'srirangam')
print(personone.greet())
persontwo = person('Deepan',27,'chennai')
print(persontwo.greet())
personthree = person('niranjan',28,'chennai')
print(personthree.greet())

# Accessing attributes
print(f'person ones name is {personone.name}')
print(f'person ones age is {personone.age}')
print(f'person ones location is {personone.location}')
print(f'person twos name is {persontwo.name}')
print(f'person twos age is {persontwo.age}')
print(f'person twos location is {persontwo.location}')
print(f'person threes name is {personthree.name}')
print(f'person threes age is {personthree.age}')
print(f'person threes location is {personthree.location}')
```

```python
class bankaccount:
  def __init__(self,account_holder,initial_balance=0):
    self.account_holder = account_holder
    self.balance =initial_balance

  def deposit(self,amount):
    if amount > 0:
      self.balance += amount
      return f'deposited ${amount}. new balance is :${self.balance}'

    else:
      return 'deposit amount must be positive'

  def withdraw(self,amount):
    if amount > 0:
      if self.balance >=amount:
        self.balance -= amount
        return f'withdrew ${amount}. new balance is :${self.balance}'
      else:
        return f'insufficient funds'
    else:
      return f'withdrawal amount must be positive'

  def check_balance(self):
    return f'account balance for {self.account_holder} is ${self.balance}'

#create a bankaccount

accountone = bankaccount('natesh',10000)
print(accountone.deposit(5000))
print(accountone.withdraw(2000))
print(accountone.check_balance())
```
